# **Data Transformation**

**Month & Day Columns**

```python
df['MonthName'] = pd.to_datetime(df['Month'], format='%m').dt.strftime('%b')
days_map = {
    1: 'Mon', 2: 'Tue', 3: 'Wed',
    4: 'Thu', 5: 'Fri', 6: 'Sat', 7: 'Sun'
}
df['Day'] = df['DayOfWeek'].map(days_map)
```

```python
df
```

**Time HHMM reformat**

```python
time_cols = ['DepTime', 'CRSDepTime', 'ArrTime', 'CRSArrTime']

for col in time_cols:
    df[col] = df[col].astype(str).str.zfill(4)
    df[col + '_hour'] = df[col].str[:2].astype(int)
```

```python
df.drop(time_cols,axis=1,inplace=True)
```

```python
df.columns
```

```python
df['DateKey'] = (
    df['Year'].astype(str) +
    df['Month'].astype(str).str.zfill(2) +
    df['DayofMonth'].astype(str).str.zfill(2)
).astype(int)

df
```

# **Data Modeling (Star Schema)**

```python
dim_date = df[['DateKey','Year', 'Month', 'DayofMonth', 'DayOfWeek', 'MonthName', 'Day']].drop_duplicates()
```

```python
dim_carrier = df[['UniqueCarrier']].drop_duplicates().reset_index(drop=True)
dim_carrier['CarrierKey'] = dim_carrier.index
df = df.merge(dim_carrier, on='UniqueCarrier', how='left')
```

```python
dim_origin = df[['Origin']].drop_duplicates().reset_index(drop=True)
dim_origin['OriginKey'] = dim_origin.index

df = df.merge(dim_origin, on='Origin', how='left')
```

```python
dim_dest = df[['Dest']].drop_duplicates().reset_index(drop=True)
dim_dest['DestKey'] = dim_dest.index

df = df.merge(dim_dest, on='Dest', how='left')
```

```python
dim_time = pd.DataFrame({
    'Hour': range(24)
})

dim_time['TimeKey'] = dim_time['Hour']

def get_period(h):
    if h < 12:
        return 'Morning'
    elif h < 18:
        return 'Afternoon'
    else:
        return 'Evening'

dim_time['Period'] = dim_time['Hour'].apply(get_period)
```

```python
dim_flight = df[['FlightNum', 'TailNum']].drop_duplicates().reset_index(drop=True)
dim_flight['FlightKey'] = dim_flight.index
df = df.merge(dim_flight, on=['FlightNum', 'TailNum'], how='left')
```

```python
df.columns
```

```python
df = df.rename(columns={
    'DepTime_hour':'DepTimeKey',
    'CRSDepTime_hour':'CRSDepTimeKey', 
    'ArrTime_hour':'ArrTimeKey', 
    'CRSArrTime_hour':'CRSArrTimeKey'
    
})
```

```python
fact_cols = [
    # FKs
    'DateKey',
    'CarrierKey',
    'OriginKey',
    'DestKey',
    'FlightKey',
    'DepTimeKey',
    'CRSDepTimeKey',
    'ArrTimeKey',
    'CRSArrTimeKey',
    
    # Measures
    'ArrDelay',
    'DepDelay',
    'ActualElapsedTime',
    'CRSElapsedTime',
    'AirTime',
    'Distance',
    'TaxiIn',
    'TaxiOut',
    'CarrierDelay',
    'WeatherDelay',
    'NASDelay',
    'SecurityDelay',
    'LateAircraftDelay',
    'Cancelled',
    'Diverted'
]

fact_table = df[fact_cols].copy()
```

```python
import os

folder = 'airline_star_schema'
os.makedirs(folder, exist_ok=True)

# Fact
fact_table.to_csv(f'{folder}/fact_flight_delays.csv', index=False)

# Dimensions
dim_date.to_csv(f'{folder}/dim_date.csv', index=False)
dim_carrier.to_csv(f'{folder}/dim_carrier.csv', index=False)
dim_origin.to_csv(f'{folder}/dim_origin_airport.csv', index=False)
dim_dest.to_csv(f'{folder}/dim_destination_airport.csv', index=False)
dim_time.to_csv(f'{folder}/dim_time.csv', index=False)
dim_flight.to_csv(f'{folder}/dim_flight.csv', index=False)
```
